```python
import gradio as gr
```

```python
def gradio_interface(img):
  image = best_model.predict(img)
  image = image[0].plot()
  return image
```

```python
check=gr.Interface(fn=gradio_interface,inputs=gr.Image("numpy"),outputs=gr.Image("numpy"))
```

```python
check.launch()
```
